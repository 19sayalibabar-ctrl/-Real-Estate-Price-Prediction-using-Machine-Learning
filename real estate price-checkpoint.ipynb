```python
import pandas as pd

```

```python
df1=pd.read_csv("Bengaluru_House_Data.csv")

df1.head()
```

```python
df1.groupby("area_type")['area_type'].agg('count')
```

```python
df2=df1.drop(['area_type','availability','society'],axis='columns')
df2.head()
```

```python
df1.groupby("size")['size'].agg('count')
```

```python
df2['BHK'] = df2['size'].str.extract('(\d+)').astype("float")

df2
```

```python
df4=df2.drop(['size'],axis='columns')
df4.head()
```

```python
df4.isnull().sum()
```

```python
df5 = df4.dropna(subset=['location'])
df5.head()
```

```python
df6 = df5.copy()  # Create a proper copy
df6['bath'].fillna(df6['bath'].median(), inplace=True)

df6.head()
```

```python
df6['balcony'].fillna(0, inplace=True)
```

```python
df6.dropna(subset=['BHK'], inplace=True)
```

```python
df6.isnull().sum()  # Should print all zeros if handled properly
```

```python
df6['BHK'].unique()
```

```python
def convert_sqft_to_num(x):
    tokens = x.split('-')
    if len(tokens) == 2:
        return (float(tokens[0])+float(tokens[1]))/2
    try:
        return float(x)
    except:
        return None 
```

```python

df6.total_sqft = df6.total_sqft.apply(convert_sqft_to_num)
df6 = df6[df6.total_sqft.notnull()]
df6.head(2)
```

```python
df7=df6.copy()
df7['price_per_sqft']=df6['price']*100000/df6['total_sqft']
```

```python
df7.head()
```

```python
len(df7['location'].unique())
```

```python
df7.location = df7.location.apply(lambda x: x.strip())
location_stats = df7['location'].value_counts(ascending=False)
location_stats
```

```python
location_stats_less_than_10 = location_stats[location_stats<=10]
location_stats_less_than_10
```

```python
df7.location = df7.location.apply(lambda x: 'other' if x in location_stats_less_than_10 else x)
len(df7.location.unique())
```
